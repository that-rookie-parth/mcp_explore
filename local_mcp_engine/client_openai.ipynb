```python
import os

from dotenv import load_dotenv
from IPython.display import Markdown, display
from langchain_mcp_adapters.client import MultiServerMCPClient
from langchain_openai.chat_models import ChatOpenAI
from langgraph.prebuilt import create_react_agent
from langchain_ollama import ChatOllama
```

```python
client = MultiServerMCPClient(
    {
        "embedded_db": {
            "url": "http://localhost:8000/mcp",
            "transport": "streamable_http",
        }
    }
)
```

```python
tools = await client.get_tools()
```

```python
load_dotenv()
```

#### Ollama: 
**Note**: 
- Not all models available in Ollama support agent and tool calls, so choose the model accordingly
- Also multiple tool calls are not always supportive for that more stronger model would be required.

```python
llm = ChatOllama(model="llama3.1:8b")
```

```python
agent = create_react_agent(model=llm, tools=tools)
```

```python
people = await agent.ainvoke(
    {
        "messages": [
            {
                "role": "user",
                "content": "Give me the information about all the people in the db.",
            }
        ]
    }
)
```

```python
content = people["messages"][-1].content
display(Markdown(content))
```

#### Uncommnet below call to make the agent to multiple tool calls.

```python
multiple_tool_call = await agent.ainvoke(
    {
        "messages": [
            {
                "role": "user",
                "content": """
                Can you do the following for me:
                1. Add information for Barry, a 21-year-old athlete to the db.
                2. Give the updated information about the people in db.
                3. Remove the information about the first person in the db.
                4. Give the final db information. 
                """
            }
        ]
    }
)
```

```python
content = multiple_tool_call["messages"][-1].content
display(Markdown(content))
```
